```python
#| export
import torch

def sincos_2d(grid_size: int, dim: int, temperature: float = 10000.):

    """Fixed 2-D sin-cos embedding, [grid_size**2, dim], row-major (q = row*grid_size + col)."""

    assert dim % 4 == 0
    ys, xs = torch.meshgrid(torch.arange(grid_size), torch.arange(grid_size), indexing='ij')
    omega = 1. / temperature ** (torch.arange(dim // 4) / (dim // 4))
    x = xs.flatten()[:, None] * omega[None]
    y = ys.flatten()[:, None] * omega[None]
    return torch.cat([x.sin(), x.cos(), y.sin(), y.cos()], dim=1)
```
